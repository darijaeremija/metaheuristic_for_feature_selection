```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score
```

```python
def evaluate_svc(code,X_train,y_train,n_splits = 5,scoring = 'accuracy'):    
    selected_features = X_train.columns[code]    
    model = SVC()
    scores = cross_val_score(model, X_train[selected_features], y_train, cv=n_splits, scoring=scoring) 
    return scores.mean()
```
