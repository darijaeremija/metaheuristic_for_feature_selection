```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score
```

```python
def evaluate_all_features_svc(X_train, y_train, X_test, y_test,n_splits = 5,scoring = 'accuracy'):
    model = SVC()
    train_score = cross_val_score(model, X_train, y_train, cv=n_splits, scoring=scoring).mean()
    model.fit(X_train, y_train)
    test_score = model.score(X_test, y_test)
   
    return train_score, test_score
```

```python
def evaluate_selected_features_svc(code,X_train,y_train,n_splits = 5,scoring = 'accuracy'):    
    selected_features = X_train.columns[code]    
    model = SVC()
    scores = cross_val_score(model, X_train[selected_features], y_train, cv=n_splits, scoring=scoring).mean()
    return scores
```

```python
def evaluate_on_test_svc(code, X_train, y_train, X_test, y_test):
    selected_features = X_train.columns[code]
    model = SVC()
    model.fit(X_train[selected_features], y_train)
    return model.score(X_test[selected_features], y_test)
```
