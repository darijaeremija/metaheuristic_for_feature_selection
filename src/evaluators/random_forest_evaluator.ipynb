```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
```

```python
def evaluate_random_forest(code,X_train,y_train,n_splits = 5,scoring = 'balanced_accuracy',n = 100):    
    selected_features = X_train.columns[code]    
    model = RandomForestClassifier(n_estimators=n)
    scores = cross_val_score(model, X_train[selected_features], y_train, cv=n_splits, scoring=scoring) 
    return scores.mean()
```

```python
from sklearn.metrics import classification_report,accuracy_score


def evaluate_test(code,X_train,y_train,X_test,y_test,n_split = 5):
    selected_features = X_train.columns[code]    
    model = RandomForestClassifier(n_estimators=100)
    model.fit(X_train[selected_features], y_train)

    y_pred = model.predict(X_test[selected_features])
    acc = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred)

    print("Test accuracy:", acc)
    print("Report:\n", report)
```
