```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
def evaluate_random_forest(code,X_train,y_train,X_validation,y_validation):    
    selected_features = X_train.columns[code]    
    model = RandomForestClassifier(n_estimators=100)
    model.fit(X_train[selected_features],y_train)
    y_predicted = model.predict(X_validation[selected_features])
    return accuracy_score(y_validation,y_predicted)    
```
