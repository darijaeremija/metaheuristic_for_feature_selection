```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
```

```python
def evaluate_all_features_knn(X_train, y_train, X_test, y_test,n_splits = 5,scoring = 'accuracy', k = 5):
    model = KNeighborsClassifier(k)
    train_score = cross_val_score(model, X_train, y_train, cv=n_splits, scoring=scoring).mean()
    model.fit(X_train, y_train)
    test_score = model.score(X_test, y_test)
   
    return train_score, test_score
```

```python
def evaluate_selected_features_knn(code,X_train,y_train,n_splits = 5,scoring = 'accuracy', k = 5):    
    selected_features = X_train.columns[code]    
    model = KNeighborsClassifier(k)
    scores = cross_val_score(model, X_train[selected_features], y_train, cv=n_splits, scoring=scoring).mean()
    return scores
```

```python
def evaluate_on_test_knn(code, X_train, y_train, X_test, y_test, k=5):
    selected_features = X_train.columns[code]
    model = KNeighborsClassifier(k)
    model.fit(X_train[selected_features], y_train)
    return model.score(X_test[selected_features], y_test)
```
