```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.metrics import accuracy_score
```

```python
def evaluate_knn(code,X_train,y_train,X_validation,y_validation, k = 5):    
    selected_features = X_train.columns[code]    
    model = KNeighborsClassifier(k)
    model.fit(X_train[selected_features],y_train)
    y_predicted = model.predict(X_validation[selected_features])
    return accuracy_score(y_validation,y_predicted)    
```
