```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
ionosphere = fetch_ucirepo(id=52) 
  
# data (as pandas dataframes) 
X = ionosphere.data.features 
y = ionosphere.data.targets 
  
# metadata 
#print(ionosphere.metadata) 
  
# variable information 
#print(ionosphere.variables) 
```

```python
y
```

```python
y.loc[:, 'Class'] = y['Class'].map({'g': 0, 'b': 1}).astype(int)
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,stratify=y,test_size=0.3)
```

```python
scaler = StandardScaler().fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train = pd.DataFrame(X_train,columns=X.columns)
X_test = pd.DataFrame(X_test,columns=X.columns)
```

```python
X_train.to_csv("../../datasets/ionosphere/X_train.csv", index=False)
X_test.to_csv("../../datasets/ionosphere/X_test.csv", index=False)

y_train.to_csv("../../datasets/ionosphere/y_train.csv", index=False)
y_test.to_csv("../../datasets/ionosphere/y_test.csv", index=False)
```
