```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("../../datasets/oral_toxicity/qsar_oral_toxicity.csv", header=None, sep = ';');
```

```python
col_names = [f"f{i}" for i in range(data.shape[1] - 1)] + ["target"]
data.columns = col_names
```

```python
data["target"] = data["target"].map({"negative": 0, "positive": 1})
```

```python
data.head()
```

```python
X = data.drop("target",axis=1)
X.head()
```

```python
y = data[["target"]]
y.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,stratify=y,test_size=0.3)
X_train,X_validation,y_train,y_validation = train_test_split(X_train,y_train,test_size = 0.3,stratify = y_train)
```

```python
scaler = StandardScaler().fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train = pd.DataFrame(X_train,columns=X.columns)
X_test = pd.DataFrame(X_test,columns=X.columns)
```

```python
X_train.to_csv("../../datasets/oral_toxicity/X_train.csv", index=False)
X_test.to_csv("../../datasets/oral_toxicity/X_test.csv", index=False)

y_train.to_csv("../../datasets/oral_toxicity/y_train.csv", index=False)
y_test.to_csv("../../datasets/oral_toxicity/y_test.csv", index=False)
```
