```python
import pandas as pd
```

```python
data = pd.read_csv("../datasets/oral_toxicity/qsar_oral_toxicity.csv", header=None, sep = ';');
```

```python
print(data.shape)
```

```python
data.head()
```

```python
col_names = [f"f{i}" for i in range(data.shape[1] - 1)] + ["target"]
data.columns = col_names
```

```python
data.head()
```

```python
data["target"] = data["target"].map({"negative": 0, "positive": 1})
```

```python
data.head()
```

```python
X = data.drop("target",axis=1)
X.head()
```

```python
y = data[["target"]]
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,stratify=y,test_size=0.2)
X_train,X_validation,y_train,y_validation = train_test_split(X_train,y_train,test_size = 0.2,stratify = y_train)
```

```python
X_train.to_csv("../datasets/oral_toxicity/X_train.csv", index=False)
X_validation.to_csv("../datasets/oral_toxicity/X_validation.csv", index=False)
X_test.to_csv("../datasets/oral_toxicity/X_test.csv", index=False)

y_train.to_csv("../datasets/oral_toxicity/y_train.csv", index=False)
y_validation.to_csv("../datasets/oral_toxicity/y_validation.csv",index = False)
y_test.to_csv("../datasets/oral_toxicity/y_test.csv", index=False)
```
