```python
import pandas as pd
from matplotlib import pyplot as plt
import openml
```

```python
def read_data():
    dataset = openml.datasets.get_dataset(dataset_id=1104,download_data=False )
    data = dataset.get_data(dataset_format='dataframe', target=dataset.default_target_attribute)
    X, y = data[0], data[1]
    y = y.map({'ALL':0,'AML':1})
    return X,pd.DataFrame(y)
```

```python
X,y = read_data()
```

```python
y.value_counts()
```

```python
#X.describe()
```

```python
counts = y.value_counts()
plt.bar(['ALL','AML'],counts)
plt.title('Class distribution')
plt.xlabel("target class")
plt.show()
```

```python
X = X.dropna()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,stratify=y,test_size=0.2)
X_train,X_validation,y_train,y_validation = train_test_split(X_train,y_train,test_size = 0.2,stratify = y_train)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler().fit(X_train)
X_train = scaler.transform(X_train)
X_validation = scaler.transform(X_validation)
X_test = scaler.transform(X_test)
```

```python
X_train = pd.DataFrame(X_train,columns=X.columns)
X_validation = pd.DataFrame(X_validation,columns=X.columns)
X_test = pd.DataFrame(X_test,columns=X.columns)
```

```python
X_train.to_csv("../datasets/leukemia/X_train.csv", index=False)
X_validation.to_csv("../datasets/leukemia/X_validation.csv", index=False)
X_test.to_csv("../datasets/leukemia/X_test.csv", index=False)

y_train.to_csv("../datasets/leukemia/y_train.csv", index=False)
y_validation.to_csv("../datasets/leukemia/y_validation.csv",index = False)
y_test.to_csv("../datasets/leukemia/y_test.csv", index=False)
```
